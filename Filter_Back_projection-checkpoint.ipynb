```python
# Defining the function f(x, y) representing the image
def f(x, y):
    # Ensure coordinates are within the image bounds
    if 0 <= x < image_array.shape[1] and 0 <= y < image_array.shape[0]:
        # Return the pixel value at the corresponding coordinates
        return image_array[int(y), int(x)]
    else:
        return 0  # Return 0 for coordinates outside the image bounds

# Test the function with some coordinates
pixel_value = f(1500, 1500)
print("Pixel value at (100, 200):", pixel_value)
```

```python
# Define the integrand function with constant radius
def integrand(z, theta):
    x = constant_radius * np.cos(theta) - z * np.sin(theta)
    y = constant_radius * np.sin(theta) + z * np.cos(theta)
    return f(x, y)

# Define the Radon transform function
def radon_transform(theta):
    return quad(integrand, -np.inf, np.inf, args=(theta))[0]
```

```python
import numpy as np
from skimage.util import view_as_windows

def extract_patches(image_array, mask_array, patch_size=(184, 184), step=184):
    """
    Extract patches from image and mask arrays.

    Parameters:
        image_array (numpy.ndarray): Input image array.
        mask_array (numpy.ndarray): Input mask array.
        patch_size (tuple): Size of the patches (height, width).
        step (int): Step size for patch extraction.

    Returns:
        image_patches (numpy.ndarray): Extracted image patches.
        mask_patches (numpy.ndarray): Extracted mask patches.
        patch_positions_array (numpy.ndarray): Array of patch positions.
    """

    # Split the image and mask arrays into patches
    image_patches = view_as_windows(image_array, patch_size, step=step)
    mask_patches = view_as_windows(mask_array, patch_size, step=step)

    # Reshape the patches to flatten them into 2D arrays
    image_patches_flat = image_patches.reshape(-1, *patch_size)
    mask_patches_flat = mask_patches.reshape(-1, *patch_size)

    # Add the color channel to the patches
    image_patches_flat = np.expand_dims(image_patches_flat, axis=3)
    mask_patches_flat = np.expand_dims(mask_patches_flat, axis=3)

    # Save the position of each patch
    patch_positions = []

    for i in range(image_patches.shape[0]):
        for j in range(image_patches.shape[1]):
            start_row = i * step
            start_col = j * step
            end_row = start_row + patch_size[0]
            end_col = start_col + patch_size[1]

            patch_positions.append({
                'start_row': start_row,
                'start_col': start_col,
                'end_row': end_row,
                'end_col': end_col
            })

    # Save the patch positions as a NumPy array
    patch_positions_array = np.array(patch_positions)

    return image_patches_flat, mask_patches_flat, patch_positions_array
```

```python
# When you set the grain to be the value 1 # background 
# You are suppose to use the calculate_pore_properties_gray_version_rock to calculate the properties
import numpy as np

def calculate_pore_properties_gray_version_rock(segmented_image, pixel_area, desired_circle_radius, circle_center):
    """
    Calculate pore properties including pore area and porosity for a grayscale segmented image within a circular region.

    Parameters:
    - segmented_image: 2D segmented image where pores are represented by values between 0 and 1.
    - pixel_area: Area of a single pixel in the image.
    - desired_circle_radius: Radius of the circular region.
    - circle_center: Coordinates of the center of the circular region (row, column).

    Returns:
    - grain_area: Total area of segmented grain regions within the circular region.
    - fraction_matrix: Matrix calculated as the ratio of grain area to the total area within the circular region.
    - porosity: Porosity within the circular region.
    - pore_area: Total area of pore regions within the circular region.
    """
    # Check if the input image is within the valid range
    if not (np.all(0 <= segmented_image) and np.all(segmented_image <= 1)):
        raise ValueError("Input image must have values between 0 and 1 for grayscale segmentation.")
    
    # Define the coordinates of the circular region
    rows, cols = np.ogrid[:segmented_image.shape[0], :segmented_image.shape[1]]
    circle_mask = (rows - circle_center[0])**2 + (cols - circle_center[1])**2 <= desired_circle_radius**2
    
    # Apply the circular mask to the segmented image
    segmented_image_circle = segmented_image.copy()
    segmented_image_circle[~circle_mask] = 0
    
    # Calculate pore area within the circular region
    grain_area = np.sum(segmented_image_circle) * pixel_area
    
    # Calculate total area within the circular region
    total_area = np.sum(circle_mask) * pixel_area
    
    # Calculate porosity and pore area within the circular region
    fraction_matrix = grain_area / total_area
    porosity = 1.0 - fraction_matrix
    pore_area = total_area - grain_area
    
    return grain_area, fraction_matrix, porosity, pore_area
```
