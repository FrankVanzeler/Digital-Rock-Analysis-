```python
import numpy as np
import matplotlib.pyplot as plt

def generate_synthetic_image(shape, num_pores):
    # Create a blank image
    image = np.zeros(shape, dtype=int)
    
    # Randomly distribute the specified number of pores
    for _ in range(num_pores):
        # Randomly choose a location for the pore
        x = np.random.randint(0, shape[0])
        y = np.random.randint(0, shape[1])
        # Set the pixel value to represent the pore
        image[x, y] = 1
    
    return image

# Example usage:
image_shape = (100, 100)  # Define the shape of the image
num_pores = 100  # Define the number of pores
synthetic_image = generate_synthetic_image(image_shape, num_pores)

# Display the generated image
plt.imshow(synthetic_image, cmap='binary')
plt.title('Synthetic Image with Controlled Pore Connectivity')
plt.colorbar()
plt.show()
```

```python
from scipy import ndimage
from skimage.segmentation import watershed
from skimage.feature import peak_local_max
from skimage import morphology

def apply_watershed_algorithm(image):
    # Compute the Euclidean distance transform of the binary image
    distance_transform = ndimage.distance_transform_edt(image)

    # Find peaks in the distance transform
    local_maxi = peak_local_max(distance_transform, indices=False, footprint=np.ones((3, 3)), labels=image)

    # Perform watershed segmentation
    markers = ndimage.label(local_maxi)[0]
    labels = watershed(-distance_transform, markers, mask=image)

    # Remove small objects and background
    labels_cleaned = morphology.remove_small_objects(labels, min_size=10)

    # Count the number of connected pores
    num_pores = np.max(labels_cleaned)

    return labels_cleaned, num_pores

# Apply watershed algorithm to the synthetic image
segmented_image, num_pores = apply_watershed_algorithm(synthetic_image)

# Display the segmented image
plt.imshow(segmented_image, cmap='viridis')
plt.title('Segmented Image with Watershed Algorithm')
plt.colorbar()
plt.show()

# Print the number of connected pores
print("Number of connected pores:", num_pores)
```
